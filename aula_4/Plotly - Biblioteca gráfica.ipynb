### Plotly
 - Plotly é uma biblioteca Python que permite construir facilmente gráficos sofisticados.
 - Possui visualização de dados interativa
 - Permite criar gráficos que podem ser manipulados em tempo real, permitindo uma exploração de dados dinâmica e imersiva
 - Documentação: https://plotly.com/python/

```python
import plotly.express as px
```

```python
#!pip install plotly==5.1.0
```

```python
df = px.data.iris()
fig = px.scatter(df, x="sepal_width", y="sepal_length", color="species",
                 size='petal_length', hover_data=['petal_width'])


fig.show()
```

```python
df = px.data.tips()
df.head()
```

```python
df.describe()
```

```python
df = px.data.tips()
fig = px.histogram(df, x="total_bill", color="sex", nbins=10)
fig.show()
```

```python
fig = px.box(df, x="day", y="total_bill", color="sex")
fig.update_traces(quartilemethod="exclusive") # or "inclusive", or "linear" by default
fig.show()
```

```python
fig = px.box(df, x="time", y="total_bill", points="all",color="sex")
fig.show()
```

```python
df = px.data.stocks()
df.head()
```

```python
df.describe()
```

```python
df = px.data.stocks()
fig = px.line(df, x="date", y=df.columns,
              hover_data={"date": "|%B %d, %Y"},
              title='custom tick labels')
fig.update_xaxes(
    dtick="M1",
    tickformat="%b\n%Y")
fig.show()
```

```python
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/plotly/datasets/master/finance-charts-apple.csv')

fig = px.line(df, x='Date', y='AAPL.High', title='Time Series with Range Slider and Selectors')

fig.update_xaxes(
    rangeslider_visible=True,
    rangeselector=dict(
        buttons=list([
            dict(count=1, label="1m", step="month", stepmode="backward"),
            dict(count=6, label="6m", step="month", stepmode="backward"),
            dict(count=1, label="YTD", step="year", stepmode="todate"),
            dict(count=1, label="1y", step="year", stepmode="backward"),
            dict(step="all")
        ])
    )
)
fig.show()
```
